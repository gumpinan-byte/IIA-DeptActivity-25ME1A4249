```python
crop_data  = {
    "crop_name": "Rice",
    "scientific_name": "Oryza sativa",
    "crop_type": "Cereal",
    "season": "Kharif",
    "ideal_temperature_celsius": 25,
    "annual_rainfall_mm": 1200,
    "soil_type": "Clayey Loam",
    "ideal_Soil_ph": 6.5,
    "growing_duration_days":120,
    "water_requirements":"High",
    "average_yield_per_hectare_kg":4000,
    "primary_nutrients":["Nitrogen","Phosphorus","Potassius"],
    "common_pests":["stem Borer","Brown Planthopper"],
    "harvesting_method":"Manual/Mechanical",
    "storage_humidity_percent":13
}
print(crop_data)
crop_data["Area"]="Eluru"
crop_data["Pesticide"]="Urea"
crop_data["primary_nutrients"] =["Calcium", "Magnesium", "Sulphur"]
crop_data["common_pests"]=["Fall Armyworm"]
print(crop_data)
```

```python
import pandas as pd

# Convert the single dictionary to a DataFrame by wrapping it in a list
df = pd.DataFrame([crop_data])

# Save the DataFrame to a CSV file
df.to_csv('crop_data.csv', index=False)

# Display the DataFrame to verify
display(df)
```

```python
# Prepare the DataFrame for MS Excel
excel_df = df.copy()

# Clean up any list columns so they display cleanly as text in Excel
for col in excel_df.columns:
    if isinstance(excel_df[col].iloc[0], list):
        excel_df[col] = excel_df[col].apply(lambda x: "; ".join(x))

# Save with 'utf-8-sig' encoding which MS Excel uses to display special characters correctly
excel_df.to_csv('crop_data_excel.csv', index=False, encoding='utf-8-sig')

# Display the cleaned DataFrame
display(excel_df)
```

```python
# Prepare the DataFrame for Google Sheets
google_sheets_df = df.copy()

# Convert list columns to clean, comma-separated strings for Google Sheets
for col in google_sheets_df.columns:
    if isinstance(google_sheets_df[col].iloc[0], list):
        google_sheets_df[col] = google_sheets_df[col].apply(lambda x: ", ".join(x))

# Save with standard UTF-8 encoding (ideal for Google Sheets upload)
google_sheets_df.to_csv('crop_data_sheets.csv', index=False, encoding='utf-8')

# Download the file to your computer
from google.colab import files
files.download('crop_data_sheets.csv')

# Display preview
display(google_sheets_df)
```
